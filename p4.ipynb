```python
import pandas as pd
data = pd.read_csv("data4.csv")
print(data)
```

```python
x = data.iloc[:, :-1]
y = data.iloc[:, -1]
```

```python
x['Age']=x['Age'].map({'Youth':1,'Middle':2,'Senior':3})
x['Income']=x['Income'].map({'High':1,'Medium':2,'Low':3})
x['Student']=x['Student'].map({'Yes':1,'No':2})
x['Creditrating']=x['Creditrating'].map({'Excellent':1,'Fair':2})
```

```python
from sklearn.naive_bayes import GaussianNB
nb=GaussianNB()
nb.fit(x,y)
print(nb.predict([[3,3,1,1]]))
print(nb.predict([[1,1,0,0]]))
print(nb.predict([[1,1,1,1]]))
```
